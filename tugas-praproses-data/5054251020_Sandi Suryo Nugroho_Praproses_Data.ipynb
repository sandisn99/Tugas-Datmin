## **1. Unduh dan Import Dataset (train.csv)** 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('train.csv')
display(df.head())
print(df.shape)
df.info()
```

## **2. Pembersihan Data**

```python
missing_percentage = df.isnull().mean() * 100
print("Persentase Nilai Hilang:")
print(missing_percentage[missing_percentage > 0].sort_values(ascending=False))

cols_to_drop = missing_percentage[missing_percentage > 50].index
print(f"\nKolom yang dihapus (>50% hilang): {list(cols_to_drop)}")
df_cleaned = df.drop(columns=cols_to_drop)

num_cols = df_cleaned.select_dtypes(include=['int64', 'float64']).columns
df_cleaned[num_cols] = df_cleaned[num_cols].fillna(df_cleaned[num_cols].median())

cat_cols = df_cleaned.select_dtypes(include=['object']).columns
for col in cat_cols:
    df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mode()[0])
    
```

## **3. Deteksi dan Penanganan Outlier**

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x=df_cleaned['GrLivArea'])
plt.title('Boxplot GrLivArea untuk Deteksi Outlier')
plt.show()

plt.figure(figsize=(10, 5))
plt.scatter(x=df_cleaned['GrLivArea'], y=df_cleaned['SalePrice'])
plt.ylabel('SalePrice')
plt.xlabel('GrLivArea')
plt.title('GrLivArea vs SalePrice')
plt.show()
```

Seringkali pada dataset rumah, rumah yang luasnya sangat besar tapi harganya murah adalah outlier yang mengganggu model. Kita bisa menyesuaikannya (menghapusnya).

```python
df_cleaned = df_cleaned.drop(df_cleaned[(df_cleaned['GrLivArea']>4000) & (df_cleaned['SalePrice']<300000)].index)
print("\nDimensi data setelah hapus outlier:", df_cleaned.shape)
```

## **4. Transformasi Data**

```python
y = df_cleaned['SalePrice']
X = df_cleaned.drop(['SalePrice', 'Id'], axis=1) # Id buang

X_num_cols = X.select_dtypes(include=['int64', 'float64']).columns
X_cat_cols = X.select_dtypes(include=['object']).columns

# 1. Standarisasi data numerik
scaler = StandardScaler()
X_num_scaled = pd.DataFrame(scaler.fit_transform(X[X_num_cols]), columns=X_num_cols, index=X.index)

# 2. One-Hot Encoding data kategorik
X_cat_encoded = pd.get_dummies(X[X_cat_cols], drop_first=True)
X_transformed = pd.concat([X_num_scaled, X_cat_encoded], axis=1)

print("Dimensi data setelah transformasi:", X_transformed.shape)
display(X_transformed.head())
```

## **5. Reduksi Dimensi**

```python
plt.figure(figsize=(12, 10))
corr_matrix = df_cleaned[num_cols].corr()
top_corr_features = corr_matrix.index[abs(corr_matrix["SalePrice"])>0.5]
sns.heatmap(df_cleaned[top_corr_features].corr(), annot=True, cmap="coolwarm")
plt.title("Heatmap Korelasi Fitur dengan SalePrice")
plt.show()

#PCA (Principal Component Analysis)
pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(X_transformed)

print(f"Jumlah komponen sebelum PCA: {X_transformed.shape[1]}")
print(f"Jumlah komponen setelah PCA (menyimpan 95% varian): {X_pca.shape[1]}")
```

## 6. **Simpan Hasil Praproses**

```python

df_final = pd.concat([X_transformed, y], axis=1)

file_name = "5054251020_Sandi-Suryo-Nugroho_Tugas_Praproses_Data.csv"
df_final.to_csv(file_name, index=False)
print(f"Dataset akhir berhasil disimpan sebagai: {file_name}")
```
